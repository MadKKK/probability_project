```python
import numpy as np
from scipy.stats import multivariate_normal
```

```python
mu = np.array([0, 0])
sigma = np.array([[1.0, 0.8],
                  [0.8, 1.0]])
my_distribution = multivariate_normal(mean = mu, cov = sigma)
samples = my_distribution.rvs(size=3)
print("my first samples:\n", samples)
```

```python
import matplotlib.pyplot as plt
```

```python
samples_large = my_distribution.rvs(size=1000)
x1 = samples_large[:, 0]
x2 = samples_large[:, 1]

plt.figure(figsize=(8, 6))
plt.scatter(x1, x2, alpha=0.5) # alpha makes the dots slightly transparent
plt.title("2D Multivariate Normal Distribution (Correlation = 0.8)")
plt.xlabel("X1")
plt.ylabel("X2")
plt.axis('equal') # Forces the x and y axes to use the exact same scale
plt.grid(True)
plt.show()
```
